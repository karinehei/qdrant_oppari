````python
import nbformat
from nbformat.v4 import new_notebook, new_markdown_cell, new_code_cell
from pathlib import Path

cells = []

# Title and overview
cells.append(new_markdown_cell("""# Mini RAG Qdrant — Notebook Versio

Tämä on **mini\_rag\_qdrant.ipynb**, suora notebooksovitus skriptistä *tools.py*.

Notebook tekee seuraavaa:

1. Käynnistää (tarvittaessa) Qdrant-kokoelman paikalliseen Qdrant-palveluun  
2. Pilkkoo esimerkkidokumentit (chunkkaus)  
3. Laskee upotteet (embeddings) kevyellä SentenceTransformers-mallilla  
4. Tallentaa vektorit Qdrantiin (HNSW + kosini)  
5. Suorittaa semanttisen haun ja rakentaa LLM:lle syötettävän promptin

> **Vaatimus:** Paikallinen Qdrant portissa 6333.  
> Käynnistys Dockerilla:
> ```bash
> docker run -p 6333:6333 -v qdrant_storage:/qdrant/storage qdrant/qdrant:latest
> ```
"""))

# Installation cell
cells.append(new_markdown_cell("## 0) Asennus"))
cells.append(new_code_cell("""# Aja tämä vain jos riippuvuudet puuttuvat.
# Poista kommentit tarvittaessa:
# !pip install --quiet qdrant-client sentence-transformers
# Valinnainen (jos haluat myös generoinnin OpenAI:n kautta):
# !pip install --quiet openai
"""))

# Imports and config
cells.append(new_markdown_cell("## 1) Tuonnit ja perusasetukset"))
cells.append(new_code_cell("""from typing import List, Dict
from qdrant_client import QdrantClient
from qdrant_client.models import Distance, VectorParams, PointStruct, SearchParams, Filter, FieldCondition, MatchValue
from sentence_transformers import SentenceTransformer
import uuid, pprint

COLLECTION = "demo_docs"
EMBED_MODEL = "sentence-transformers/all-MiniLM-L6-v2"  # kevyt ja nopea
pp = pprint.PrettyPrinter(indent=2)
"""))

# Demo docs
cells.append(new_markdown_cell("## 2) Esimerkkidokumentit (voit korvata omillasi)"))
cells.append(new_code_cell("""DOCS = [
    {"id": "polku/1", "title": "Qdrant pähkinänkuoressa",
     "text": "Qdrant on vektoridatabäsi, joka tukee HNSW-indeksointia, metadatafilttereitä ja reaaliaikaisia päivityksiä."},
    {"id": "polku/2", "title": "RAG-perusteet",
     "text": "RAG yhdistää haun ja generoinnin: dokumentit pilkotaan, indeksoidaan vektorimuotoon ja haetaan semanttisesti."},
    {"id": "polku/3", "title": "Embeddings lyhyesti",
     "text": "Upotteet (embeddings) kuvaavat tekstin merkityssisältöä, mahdollistaen samankaltaisuushakuja kosinietäisyydellä."},
]"""))

# Helper functions
cells.append(new_markdown_cell("## 3) Aputoiminnot"))
cells.append(new_code_cell("""def chunk(text: str, max_len: int = 256) -> List[str]:
    \"\"\"Erittäin yksinkertainen chunkkaus lausejakajalla.\"\"\"
    s = [p.strip() for p in text.replace("\\n", " ").split(".") if p.strip()]
    chunks, buf = [], ""
    for sent in s:
        if len(buf) + len(sent) + 1 <= max_len:
            buf = f"{buf} {sent}".strip()
        else:
            if buf: chunks.append(buf)
            buf = sent
    if buf: chunks.append(buf)
    return chunks

def build_corpus(docs: List[Dict]) -> List[Dict]:
    corpus = []
    for d in docs:
        for i, ch in enumerate(chunk(d["text"])):
            corpus.append({
                "uid": f"{d['id']}#{i}",
                "title": d["title"],
                "text": ch,
                "source": d["id"],
                "domain": "demo"  # esimerkki metadatasta
            })
    return corpus

def ensure_collection(client: QdrantClient, dim: int):
    \"\"\"Luo kokoelma HNSW:llä jos puuttuu (COSINE-etäisyys).\"\"\"
    exists = client.collection_exists(COLLECTION)
    if not exists:
        client.recreate_collection(
            collection_name=COLLECTION,
            vectors_config=VectorParams(size=dim, distance=Distance.COSINE),
        )

def index_points(client: QdrantClient, model: SentenceTransformer, corpus: List[Dict]):
    vectors = model.encode([c["text"] for c in corpus], normalize_embeddings=True).tolist()
    points = []
    for v, c in zip(vectors, corpus):
        points.append(
            PointStruct(
                id=str(uuid.uuid4()),
                vector=v,
                payload={
                    "title": c["title"],
                    "text": c["text"],
                    "source": c["source"],
                    "domain": c["domain"]
                }
            )
        )
    client.upsert(collection_name=COLLECTION, points=points)

def semantic_search(client: QdrantClient, model: SentenceTransformer, query: str, top_k: int = 4, hnsw_ef: int = 128, domain: str = None):
    q_vec = model.encode([query], normalize_embeddings=True).tolist()[0]
    search_params = SearchParams(hnsw_ef=hnsw_ef)
    q_filter = None
    if domain:
        q_filter = Filter(
            must=[FieldCondition(key="domain", match=MatchValue(value=domain))]
        )
    hits = client.search(
        collection_name=COLLECTION,
        query_vector=q_vec,
        limit=top_k,
        search_params=search_params,
        query_filter=q_filter
    )
    return [{
        "score": round(h.score, 3),
        "title": h.payload.get("title"),
        "text": h.payload.get("text"),
        "source": h.payload.get("source"),
        "domain": h.payload.get("domain"),
    } for h in hits]

def build_prompt(question: str, contexts: List[Dict]) -> str:
    ctx = "\\n\\n---\\n\\n".join([f"({c['score']}) {c['title']} — {c['text']}" for c in contexts])
    return f\"\"\"Vastaa kysymykseen hyödyntäen vain alla olevaa kontekstia. Jos vastaus ei löydy, sano 'en tiedä' ja pyydä lisää lähteitä.

KYSYMYS:
{question}

KONTEKSTI:
{ctx}
\"\"\""""))

# Initialize model and client
cells.append(new_markdown_cell("## 4) Malli ja Qdrant-client"))
cells.append(new_code_cell("""model = SentenceTransformer(EMBED_MODEL)
client = QdrantClient(url="http://localhost:6333")
print("Embedding-dimension:", model.get_sentence_embedding_dimension())
print("Yhteys Qdrantiin OK")"""))

# Build corpus and ensure collection
cells.append(new_markdown_cell("## 5) Korpuksen luonti ja kokoelman varmistus"))
cells.append(new_code_cell("""corpus = build_corpus(DOCS)
ensure_collection(client, dim=model.get_sentence_embedding_dimension())
len(corpus)"""))

# Index points
cells.append(new_markdown_cell("## 6) Indeksointi Qdranttiin"))
cells.append(new_code_cell("""index_points(client, model, corpus)
print("Indeksoitu pisteitä:", len(corpus))"""))

# Query cell
cells.append(new_markdown_cell("## 7) Haku — kirjoita kysymys tähän"))
cells.append(new_code_cell("""question = "Miksi vektoridatabase on hyödyllinen RAG-järjestelmässä?"
results = semantic_search(client, model, question, top_k=3, hnsw_ef=128, domain="demo")
print("== Haun osumat ==")
for r in results:
    print(f"{r['score']}: {r['title']} [{r['source']}] -> {r['text'][:120]}...")
results"""))

# Build prompt cell
cells.append(new_markdown_cell("## 8) LLM-promptin rakentaminen"))
cells.append(new_code_cell("""prompt = build_prompt(question, results)
print(prompt)"""))

# Optional OpenAI generation cell
cells.append(new_markdown_cell("## 9) (Valinnainen) Generointi OpenAI-rajapinnalla"))
cells.append(new_code_cell("""# Poista kommentit ja aseta avain, jos haluat generoinnin suoraan notebookista.
# from openai import OpenAI
# import os
# os.environ["OPENAI_API_KEY"] = "PAHVI-TÄHÄN"
# client_ = OpenAI()
# completion = client_.chat.completions.create(
#     model="gpt-4o-mini",
#     messages=[{"role":"user","content":prompt}]
# )
# print("\\n== Mallin vastaus ==\\n")
# print(completion.choices[0].message.content)
"""))

# Tips / next steps
cells.append(new_markdown_cell("""## 10) Pieniä parannuksia
- **Parempi recall:** nosta `hnsw_ef` arvoon 256–512 (nostaa viivettä hieman).  
- **Hybridi-haku:** yhdistä BM25 (esim. Elasticsearch) + vektorit ja tee yhdistetty rankkaus.  
- **Filtterit:** lisää payloadiin kenttiä kuten `doc_type`, `language` ja käytä `query_filter`-ehtoja.  
- **Omat datat:** korvaa `DOCS` lukemalla tekstit tiedostoista tai tietokannasta.  
"""))

nb = new_notebook(cells=cells, metadata={"kernelspec": {"name": "python3", "display_name": "Python 3"}})

out_path = Path("/mnt/data/mini_rag_qdrant.ipynb")
out_path.parent.mkdir(parents=True, exist_ok=True)
nbformat.write(nb, str(out_path))

str(out_path)
````
